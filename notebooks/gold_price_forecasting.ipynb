````python
# Dataset Splitting:
# X, y Separation and Train-Test split:

import pandas as pd
from IPython.display import HTML, display
from sklearn.model_selection import train_test_split


def split_and_display(
    df,
    target_col,
    test_size=0.2,
    random_state=42,
    head_n=5,
    show_html=True,
    **kwargs,
):
    """Splits a pandas DataFrame into features (X) and target (y), performs a train-test

    split, and renders an HTML visualization preview for Jupyter Notebooks.

    Parameters:
    -----------
    df : pandas.DataFrame
        The full dataset (without the Target col) to split.
    target_col : str
        Name of the target column to predict.
    test_size : float or int, default=0.2
        Proportion or absolute number of the dataset to include in the test
        split.
    random_state : int, default=42
        Controls the shuffling applied to the data before applying the split.
    head_n : int, default=5
        Number of rows to show in HTML previews.
    show_html : bool, default=True
        Whether to display the IPython HTML output directly in the notebook.
    **kwargs :
        Additional keyword arguments passed to `sklearn.model_selection.train_test_split`.

    Returns:
    --------
    X_train, X_test, y_train, y_test : DataFrames/Series tuple

    Usage Example:
    -------------
    # Simple usage with default 80:20 split:
    X_train, X_test, y_train, y_test = split_and_display(
        df=df, target_col="House_Price"
    )

    # Custom parameters (e.g., stratified classification split or 70:30 ratio):
    X_train, X_test, y_train, y_test = split_and_display(
        df=df, target_col="House_Price", test_size=0.3, stratify=df["House_Price"]
    )
    """
    # 1. Feature-Target Separation
    X = df.drop(columns=[target_col])
    y = df[target_col]

    # 2. Train-Test Split
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state, **kwargs)

    # 3. HTML Visualization Construction
    if show_html:
        # Convert y Series to DataFrames safely for proper .to_html() rendering
        y_frame = y.to_frame()
        y_train_frame = y_train.to_frame()
        y_test_frame = y_test.to_frame()

        html_str_X_y = f"""
        <div style='margin-bottom: 30px; overflow: auto;'>
          <div style='float: left; margin-right: 40px;'>
            <h4>X, y Separation:</h4>
            <h4>Features (X.head)</h4>
            {X.head(head_n).to_html()}
          </div>
          <div style='float: left;'>
            <h4>Target (y.head)</h4>
            {y_frame.head(head_n).to_html()}
          </div>
        </div>
        """

        html_splits = f"""
        <div style='margin-bottom:30px;'>
          <!-- Train Row -->
          <div style='display:flex; gap:40px; margin-bottom:30px; align-items:flex-start;'>
            <div>
              <h4>TrainSet: Features (X_train.head)</h4>
              {X_train.head(head_n).to_html()}
            </div>
            <div>
              <h4>TrainSet: Target (y_train.head)</h4>
              {y_train_frame.head(head_n).to_html()}
            </div>
          </div>
          <!-- Test Row -->
          <div style='display:flex; gap:40px; align-items:flex-start;'>
            <div>
              <h4>TestSet: Features (X_test.head)</h4>
              {X_test.head(head_n).to_html()}
            </div>
            <div>
              <h4>TestSet: Target (y_test.head)</h4>
              {y_test_frame.head(head_n).to_html()}
            </div>
          </div>
        </div>
        """

        html_str_train_test_shape = f"""
        <div style='border-top: 1px solid #ccc; padding-top: 10px;'>
          <h3>Dataset Dimensions Summary</h3>
          <p><b>X_train:</b> {X_train.shape[0]} rows × {X_train.shape[1]} columns</p>
          <p><b>y_train:</b> {y_train.shape[0]} elements</p>
          <p><b>X_test:</b> {X_test.shape[0]} rows × {X_test.shape[1]} columns</p>
          <p><b>y_test:</b> {y_test.shape[0]} elements</p>
        </div>
        """

        display(HTML(html_str_X_y))
        display(HTML(html_splits))
        display(HTML(html_str_train_test_shape))

    return X_train, X_test, y_train, y_test

# My note: stratify=y is used for Classification problems, but in Regression problem, if target col is highly skewed only then we may need binned stratification
# but `Gold_T+22` is almost normally distributed so no actions needed

X_train, X_test, y_train, y_test = split_and_display(df=df, target_col='Gold_T+22')
````

````python
"""
# My Note:
1. Outliear handling gulo implement korte hoy custom class e. not with `FunctionTransformer`. cause otherwise 'Data Leakage' hote pare!
jokhon e bhabbo: train er calculted ta test e use korte hobe, tokhoni Class banano instead of custom Transformer function!!!

2. Pipeline e Normally dist/Skewed whatever the cols are, you must implement all the general steps (Imputation, Outliear, Encode, Scale etc as needed) for Robustness
"""

# Custom Classes and Functions:
from sklearn.base import BaseEstimator, TransformerMixin
class ZScoreOutlierClipper(BaseEstimator, TransformerMixin):
    def __init__(self, threshold=3.0):
        self.threshold = threshold
        self.lower_bounds_ = None
        self.upper_bounds_ = None

    def fit(self, X, y=None):
        X_arr = np.asarray(X)
        means = np.mean(X_arr, axis=0)
        stds = np.std(X_arr, axis=0)

        # Preventing division by zero
        stds = np.where(stds == 0, 1e-8, stds)

        self.lower_bounds_ = means - (self.threshold * stds)
        self.upper_bounds_ = means + (self.threshold * stds)
        return self

    def transform(self, X):
        X_arr = np.asarray(X).copy()
        # Vectorized clipping instead of a Python loop (Faster!)
        return np.clip(X_arr, self.lower_bounds_, self.upper_bounds_)


# Preprocessing Pipelines:
p1 = Pipeline(  # for Normally dist Cols
    steps=[
        ('imputer', SimpleImputer(strategy='mean')), # NOT NEED  for the Dataset, but keeping Imputation step just for ROBUSTNESS!
        ('outliear_handling',ZScoreOutlierClipper(threshold=3.0)), # Z-score based outliear handling # WARNING! Outliear handling gulo implement korte hoy custom class e. not with `FunctionTransformer`. cause otherwise 'Data Leakage' hote pare! jokhon e bhabbo: train er calculted ta test e use korte hobe, tokhoni Class banano instead of custom Transformer function!!!
        ('scaling', StandardScaler())
    ]
)

from sklearn.preprocessing import PowerTransformer
p2 = Pipeline(  # for Skewed Cols
    steps=[
        ('imputer', SimpleImputer(strategy='median')), # NOT NEED  for the Dataset, but keeping Imputation step just for ROBUSTNESS!
        ('log_transform', PowerTransformer(method='yeo-johnson', standardize=True)) # handles outliears and scales. [? Skewed to --> normall dist convertion]
    ]
)

# columntransformer:
preprocessor = ColumnTransformer(
    transformers=[
        ('norm_dist_col', p1, normally_dist_cols),
        ('skewed_col', p2, highly_skewed_cols)
    ],
    remainder='drop',
    verbose_feature_names_out=False
)

display(preprocessor)
````

````python
# Custom Functions and Classes:

# Feature vs Feature combined with Feature vs Target:
from sklearn.base import BaseEstimator, TransformerMixin
class CollinearFeatureDropper(BaseEstimator, TransformerMixin):
    """
    Identifies pairwise correlated features. When a pair exceeds
    the threshold, it compares their mutual information scores
    against the regression target and drops the weaker predictor.
    """

    def __init__(self, threshold=0.85, random_state=42):
        self.threshold = threshold
        self.random_state = random_state
        self.to_drop_ = []

    def fit(self, X, y=None):
        X_df = pd.DataFrame(X)

        # 1. Compute Feature-vs-Feature correlation matrix
        corr_matrix = X_df.corr().abs()

        # 2. Compute Feature-vs-Target scores using mutual information
        if y is not None:
            mi_scores = mutual_info_regression(
                X_df,
                y,
                random_state=self.random_state
            )

            # Replace NaNs/Infs with 0 just in case
            mi_scores = np.nan_to_num(
                mi_scores,
                nan=0.0,
                posinf=0.0,
                neginf=0.0
            )
        else:
            # Without a target, use blind dropping
            mi_scores = np.zeros(X_df.shape[1])

        # Map feature names to their MI scores
        mi_scores = pd.Series(mi_scores, index=X_df.columns)

        to_drop = set()
        columns = corr_matrix.columns

        # 3. Compare correlated feature pairs
        for i in range(len(columns)):
            for j in range(i):
                if corr_matrix.iloc[i, j] > self.threshold:
                    feat_1 = columns[i]
                    feat_2 = columns[j]

                    if feat_1 in to_drop or feat_2 in to_drop:
                        continue

                    # Drop the feature with lower mutual information
                    if mi_scores[feat_1] < mi_scores[feat_2]:
                        to_drop.add(feat_1)
                    else:
                        to_drop.add(feat_2)

        self.to_drop_ = list(to_drop)
        self.feature_names_in_ = np.asarray(X_df.columns)
        return self

    def transform(self, X):
        X_df = pd.DataFrame(X)

        # Drop weaker redundant features
        return X_df.drop(columns=self.to_drop_).values
````

**Why:** _(Collected)_
> **The Overlapping Label Leakage Trap**
> If the target col _(i.e. Gold_T+22)_ is created using a rolling or forward-looking window (e.g., predicting the price 22 days ahead from overlapping historical rows), then **standard `TimeSeriesSplit` is not enough**.
> Consecutive rows in my gold.csv dataset share massive data overlap. The training fold immediately preceding the test fold will contain samples that overlap heavily in time with the start of the test set. This introduces subtle **look-ahead bias/data leakage**, causing the cross-validation scores to look overly optimistic compared to true out-of-time deployment.

**Recommended Fixes for Financial Time-Series:**

1. **Add a Purging/Embarcation Gap:** Introduce a gap (purge) between your training set and validation/test set equal to your forecast horizon ($22$ periods) so that training samples do not share overlapping future information with the validation fold.
2. **Stationarity Check:** Ensure your features and targets are stationary (e.g., using log returns or percentage changes rather than raw prices), as non-stationary financial data often leads to spurious correlations.

**2. PCA (Principal Component Analysis)**

**Intuition:** If we imagine taking a 3D object, like a Pencil-box, and shining a flashlight on a wall to project its 2D shadow. If we angle the light right, we catch the entire shape— the spout, handle, and the body— in a single flat image without losing essential detail. PCA does this for high-dimensional data: it rotates and projects complex feature space onto new, perpendicular axes (**Principal Components**) that capture the maximum variance in as few dimensions as possible.

**Necessity:** High-dimensional data causes the **curse of dimensionality**, inflating computation costs, storage requirements, and model training times. PCA compresses dimensions while preserving maximum data variance, eliminating redundant collinear features.

**Note:** Starts from `PC1` to `PCn`, where first few PC's hold max data variance and diminishes till `PCn`.

```python
from sklearn.decomposition import PCA

X, _ = load_iris(return_X_y=True)

# Compresses 4 original raw features down to 2 principal components:

X_reduced = pca.fit_transform(X)

print(f"Original shape: {X.shape} -> Reduced shape: {X_reduced.shape}")
print(f"Explained Variance Ratio: {pca.explained_variance_ratio_.sum():.2f}")
```
